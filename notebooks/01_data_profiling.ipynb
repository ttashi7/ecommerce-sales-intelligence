```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
data_path = Path("../data/raw")
data_path
```

```python
csv_files = sorted(data_path.glob("*.csv"))

for file in csv_files:
    print(file.name)
```

```python
datasets = {}

for file in csv_files:
    datasets[file.stem] = pd.read_csv(file)

print(f"Loaded {len(datasets)} datasets.")
```

```python
for name, df in datasets.items():
    print("=" * 60)
    print(name)
    print(df.shape)
```

```python
def profile_dataframe(df):
    """Return a quick profile of a dataframe."""

    profile = {
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": df.isna().sum().sum(),
        "Duplicate Rows": df.duplicated().sum()
    }

    return profile
```

```python
summary = {}

for name, df in datasets.items():
    summary[name] = profile_dataframe(df)

summary_df = pd.DataFrame(summary).T

summary_df
```

```python
summary_df = summary_df.sort_index()

summary_df
```

```python
def inspect_table(df):

    display(df.head())

    print("\nShape:")
    print(df.shape)

    print("\nData Types")
    display(df.dtypes)

    print("\nMissing Values")
    display(df.isnull().sum())

    print("\nUnique Values")
    display(df.nunique())
```

```python
inspect_table(datasets["olist_orders_dataset"])
```

# Data Quality Assessment

```python
quality_report = []

for name, df in datasets.items():

    quality_report.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Cells": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum(),
        "Memory (MB)": round(df.memory_usage(deep=True).sum()/1024**2,2)
    })

quality_df = pd.DataFrame(quality_report)

quality_df
```

# Missing Value Analysis

```python
for name, df in datasets.items():

    missing = (
        df.isnull()
          .sum()
          .sort_values(ascending=False)
    )

    missing = missing[missing > 0]

    if len(missing):

        print("\n")
        print("="*60)
        print(name)

        display(missing)
```

# Candidate Primary Keys

## Objective

Before designing a relational database or writing SQL joins, it is important to identify columns that uniquely identify each record. These candidate primary keys will later be used to define table relationships and validate data integrity.

```python
def find_candidate_keys(df):

    candidates = []

    for column in df.columns:

        if df[column].is_unique:
            candidates.append(column)

    return candidates
```

```python
for name, df in datasets.items():

    print(name)

    print(find_candidate_keys(df))

    print()
```
